```python
# This pathological instance has been generated by AI
from core.instance import Instance, Student, Project
from core.matching import Matching

A, B, C, D = 100, 101, 102, 103

students = tuple(Student(id=i) for i in range(6))

projects = (
    Project(id=A, capacity=2),   
    Project(id=B, capacity=1),  
    Project(id=C, capacity=2),   
    Project(id=D, capacity=1),   
)

preferences = {
    0: [A, B, C, D],
    1: [A, B, D, C],
    2: [B, A, C, D],
    3: [B, A, D, C],
    4: [C, D, A, B],   
    5: [D, C, B, A],   
}


school_priorities = {
    A: [2, 3, 0, 1, 4, 5],   
    B: [0, 1, 4, 5, 2, 3],   
    C: [5, 4, 1, 0, 3, 2],
    D: [4, 5, 2, 3, 0, 1],
}

instance = Instance(
    students=students,
    projects=projects,
    preferences=preferences,
    school_priorities=school_priorities,
)
```

```python
from mechanisms.gale_shapley import gale_shapley
from mechanisms.ttc_from_matching import (
    ttc_min_rank_sum, ttc_min_borda_welfare,
    ttc_min_nash_welfare, ttc_min_max_rank,
    ttc_min_square_rank_sum, ttc_min_top_n,
)
from metrics.cost import rank_sum_cost, borda_welfare_cost, max_rank_cost, nash_welfare_cost

gs = gale_shapley(instance)

results = {
    "GS":          gs,
    "TTC rank_sum":    ttc_min_rank_sum(gs),
    "TTC borda":       ttc_min_borda_welfare(gs),
    "TTC nash":        ttc_min_nash_welfare(gs),
    "TTC max_rank":    ttc_min_max_rank(gs),
    "TTC sq_rank":     ttc_min_square_rank_sum(gs),
    "TTC top2":        ttc_min_top_n(gs, n=2),
}

metrics = {
    "rank_sum":   rank_sum_cost,
    "borda":      borda_welfare_cost,
    "max_rank":   max_rank_cost,
    "nash":       nash_welfare_cost,
}

print(f"{'':20}", "  ".join(f"{m:>10}" for m in metrics))
for name, m in results.items():
    scores = "  ".join(f"{fn(m):>10.4f}" for fn in metrics.values())
    print(f"{name:20}  {scores}")
```
